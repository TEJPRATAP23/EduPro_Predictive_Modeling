```python
import pandas as pd
import numpy as np

from pathlib import Path
import joblib
```

```python
project_folder = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling"
)

model_folder = project_folder / "models"
app_folder = project_folder / "app"
notebook_folder = project_folder / "notebooks"

revenue_model_path = (
    model_folder /
    "EduPro_Final_Revenue_Model.joblib"
)

app_file_path = (
    app_folder /
    "app.py"
)

print("Project:", project_folder)
print("Revenue model:", revenue_model_path)
print("App:", app_file_path)
```

## Load Revenue Model

```python
revenue_model = joblib.load(
    revenue_model_path
)   

print(type(revenue_model))  
```

## Define Revenue Model

```python
revenue_features = [
    "CourseCategory",
    "CourseType",
    "CourseLevel",
    "CoursePrice",
    "CourseDuration",
    "CourseRating",
    "TeacherRating",
    "YearsOfExperience",
    "Expertise"
]

print("Revenue features:", len(revenue_features))
```

## Rvenue Prediction Function

```python
def predict_revenue(course_input):

    revenue_input = course_input[
        revenue_features
    ].copy()

    prediction = revenue_model.predict(
        revenue_input
    )[0]

    prediction = max(0, prediction)

    return prediction
```

## Revenue Classification

```python
def classify_revenue(revenue):

    if revenue < 10000:
        return "Low"

    elif revenue < 25000:
        return "Moderate"

    elif revenue < 50000:
        return "High"

    else:
        return "Very High"
```

## Revenue Recommendation

```python
def generate_revenue_recommendation(
    revenue_level
):

    if revenue_level == "Low":
        return (
            "Review pricing, course positioning, "
            "target audience and promotional strategy."
        )

    elif revenue_level == "Moderate":
        return (
            "Consider targeted marketing and "
            "monitor early revenue performance."
        )

    elif revenue_level == "High":
        return (
            "Strong revenue opportunity. "
            "Prioritize launch and capacity planning."
        )

    else:
        return (
            "Very strong predicted revenue potential. "
            "Prioritize launch and prepare sufficient capacity."
        )
```

## Test Revenue Prediction

```python
test_course = pd.DataFrame([
    {
        "CourseCategory": "Technology",
        "CourseType": "Online",
        "CourseLevel": "Intermediate",
        "CoursePrice": 490.9,
        "CourseDuration": 7.55,
        "CourseRating": 4.55,
        "TeacherRating": 4.58,
        "YearsOfExperience": 24,
        "Expertise": "Programming"
    }
])
```

```python
predicted_revenue = predict_revenue(
    test_course
)

revenue_level = classify_revenue(
    predicted_revenue
)

recommendation = (
    generate_revenue_recommendation(
        revenue_level
    )
)

print("Predicted Revenue:")
print(round(predicted_revenue, 2))

print("\nRevenue Potential:")
print(revenue_level)

print("\nRecommendation:")
print(recommendation)
```

## Revenue-per-Enrollment Test

```python
predicted_enrollment = 168.23
```

```python
revenue_per_enrollment = (
    predicted_revenue /
    predicted_enrollment
)

print(
    "Revenue per expected enrollment:",
    round(revenue_per_enrollment, 2)
)
```
