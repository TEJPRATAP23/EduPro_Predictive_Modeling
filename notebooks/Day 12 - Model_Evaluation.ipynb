```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
day9_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data\EduPro_Day9_Train_Test_Splits.xlsx"
)

day10_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data\EduPro_Day10_Baseline_Models.xlsx"
)

day11_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data\EduPro_Day11_Advanced_Models.xlsx"
)

model_folder = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data"
)

model_folder.mkdir(
    parents=True,
    exist_ok=True
)

day12_file_path = (
    model_folder /
    "EduPro_Day12_Model_Evaluation.xlsx"
)

print("Day 9 input:")
print(day9_file_path)

print("\nDay 10 input:")
print(day10_file_path)

print("\nDay 11 input:")
print(day11_file_path)

print("\nDay 12 output:")
print(day12_file_path)
```

## Load Day 10 results 

```python
day10_metrics = pd.read_excel(
    day10_file_path,
    sheet_name="Baseline_Metrics"
)

day10_predictions = pd.read_excel(
    day10_file_path,
    sheet_name="Predictions"
)

print("Day 10 results loaded successfully.")

print("\nDay 10 metrics:")
print(day10_metrics)

print("\nDay 10 predictions shape:")
print(day10_predictions.shape)
```

## Load Day 11 results

```python
day11_metrics = pd.read_excel(
    day11_file_path,
    sheet_name="Advanced_Metrics"
)

day11_all_models = pd.read_excel(
    day11_file_path,
    sheet_name="All_Model_Results"
)

day11_best_models = pd.read_excel(
    day11_file_path,
    sheet_name="Best_Models"
)

day11_predictions = pd.read_excel(
    day11_file_path,
    sheet_name="Predictions"
)

print("Day 11 results loaded successfully.")

print("\nDay 11 advanced metrics:")
print(day11_metrics)

print("\nDay 11 best models:")
print(day11_best_models)

print("\nDay 11 predictions:")
print(day11_predictions.shape)
```

## Evaluation 1 — Combine all model metrics

```python
all_model_results = pd.concat(
    [
        day10_metrics,
        day11_metrics
    ],
    ignore_index=True
)

print("========== ALL MODEL METRICS ==========")

all_model_results
```

## Evaluation 2 — Rank models

```python
ranked_models = (
    all_model_results
    .sort_values(
        by=["Target", "MAE"]
    )
    .reset_index(drop=True)
)

ranked_models
```

### Add model rankings

```python
ranked_models["MAE_Rank"] = (
    ranked_models
    .groupby("Target")["MAE"]
    .rank(
        method="min",
        ascending=True
    )
)

ranked_models["RMSE_Rank"] = (
    ranked_models
    .groupby("Target")["RMSE"]
    .rank(
        method="min",
        ascending=True
    )
)

ranked_models["R2_Rank"] = (
    ranked_models
    .groupby("Target")["R2"]
    .rank(
        method="min",
        ascending=False
    )
)

ranked_models
```

## Evaluation 3 — Compare against Mean Baseline

```python
baseline_results = (
    all_model_results[
        all_model_results["Model"] == "Mean Baseline"
    ][
        ["Target", "MAE", "RMSE", "R2"]
    ]
    .rename(
        columns={
            "MAE": "Baseline_MAE",
            "RMSE": "Baseline_RMSE",
            "R2": "Baseline_R2"
        }
    )
)

baseline_results
```

### Calculate improvement

```python
evaluation_comparison = all_model_results.merge(
    baseline_results,
    on="Target",
    how="left"
)

evaluation_comparison["MAE_Improvement"] = (
    evaluation_comparison["Baseline_MAE"]
    - evaluation_comparison["MAE"]
)

evaluation_comparison["RMSE_Improvement"] = (
    evaluation_comparison["Baseline_RMSE"]
    - evaluation_comparison["RMSE"]
)

evaluation_comparison["MAE_Improvement_%"] = (
    evaluation_comparison["MAE_Improvement"]
    / evaluation_comparison["Baseline_MAE"]
) * 100

evaluation_comparison["RMSE_Improvement_%"] = (
    evaluation_comparison["RMSE_Improvement"]
    / evaluation_comparison["Baseline_RMSE"]
) * 100

evaluation_comparison.round(3)
```

## Evaluation 4 — Identify models that actually beat baseline

```python
evaluation_comparison["Beats_Baseline"] = (
    evaluation_comparison["MAE_Improvement"] > 0
)

evaluation_comparison[
    [
        "Target",
        "Model",
        "MAE",
        "Baseline_MAE",
        "MAE_Improvement",
        "MAE_Improvement_%",
        "Beats_Baseline"
    ]
].round(3)
```

## Evaluation 5 — Select best predictive model

```python
best_evaluated_models = []

for target in ["EnrollmentCount", "CourseRevenue"]:

    target_results = evaluation_comparison[
        evaluation_comparison["Target"] == target
    ].copy()

    predictive_models = target_results[
        target_results["Model"] != "Mean Baseline"
    ]

    best_row = predictive_models.loc[
        predictive_models["MAE"].idxmin()
    ]

    best_evaluated_models.append(
        {
            "Target": target,
            "Best_Predictive_Model": best_row["Model"],
            "MAE": best_row["MAE"],
            "RMSE": best_row["RMSE"],
            "R2": best_row["R2"],
            "Beats_Baseline": best_row["Beats_Baseline"],
            "MAE_Improvement_%": best_row["MAE_Improvement_%"]
        }
    )

best_predictive_models_df = pd.DataFrame(
    best_evaluated_models
)

best_predictive_models_df.round(3)
```

## Evaluation 6 — Actual vs predicted

```python
prediction_evaluation = day11_predictions.copy()

print("Prediction evaluation dataset:")
print(prediction_evaluation.shape)

prediction_evaluation.head()
```

## Evaluation 7 — Calculate prediction errors

```python
prediction_evaluation["RF_Enrollment_Error"] = (
    prediction_evaluation["Random_Forest_EnrollmentCount"]
    - prediction_evaluation["Actual_EnrollmentCount"]
)

prediction_evaluation["GB_Enrollment_Error"] = (
    prediction_evaluation["Gradient_Boosting_EnrollmentCount"]
    - prediction_evaluation["Actual_EnrollmentCount"]
)

prediction_evaluation["RF_Revenue_Error"] = (
    prediction_evaluation["Random_Forest_CourseRevenue"]
    - prediction_evaluation["Actual_CourseRevenue"]
)

prediction_evaluation["GB_Revenue_Error"] = (
    prediction_evaluation["Gradient_Boosting_CourseRevenue"]
    - prediction_evaluation["Actual_CourseRevenue"]
)

prediction_evaluation.round(2)
```

## Evaluation 8 — Absolute errors

```python
prediction_evaluation["RF_Enrollment_Absolute_Error"] = (
    prediction_evaluation["RF_Enrollment_Error"].abs()
)

prediction_evaluation["GB_Enrollment_Absolute_Error"] = (
    prediction_evaluation["GB_Enrollment_Error"].abs()
)

prediction_evaluation["RF_Revenue_Absolute_Error"] = (
    prediction_evaluation["RF_Revenue_Error"].abs()
)

prediction_evaluation["GB_Revenue_Absolute_Error"] = (
    prediction_evaluation["GB_Revenue_Error"].abs()
)

prediction_evaluation.round(2)
```

## Evaluation 9 — Revenue zero-value evaluation

```python
revenue_zero_evaluation = prediction_evaluation[
    prediction_evaluation["Actual_CourseRevenue"] == 0
].copy()

revenue_nonzero_evaluation = prediction_evaluation[
    prediction_evaluation["Actual_CourseRevenue"] > 0
].copy()

print("Zero-revenue test observations:")
print(len(revenue_zero_evaluation))

print("\nNon-zero-revenue test observations:")
print(len(revenue_nonzero_evaluation))
```

```python
print("========== ZERO REVENUE PREDICTIONS ==========")

revenue_zero_evaluation[
    [
        "CourseID",
        "CourseName",
        "Actual_CourseRevenue",
        "Random_Forest_CourseRevenue",
        "Gradient_Boosting_CourseRevenue"
    ]
].round(2)
```

## Evaluation 10 — Non-zero revenue performance

```python
nonzero_revenue_metrics = []

for model_name, prediction_column in [
    (
        "Random Forest",
        "Random_Forest_CourseRevenue"
    ),
    (
        "Gradient Boosting",
        "Gradient_Boosting_CourseRevenue"
    )
]:

    actual = revenue_nonzero_evaluation[
        "Actual_CourseRevenue"
    ]

    predictions = revenue_nonzero_evaluation[
        prediction_column
    ]

    nonzero_revenue_metrics.append(
        {
            "Model": model_name,
            "MAE": mean_absolute_error(
                actual,
                predictions
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    actual,
                    predictions
                )
            ),
            "R2": r2_score(
                actual,
                predictions
            )
        }
    )

nonzero_revenue_metrics_df = pd.DataFrame(
    nonzero_revenue_metrics
)

nonzero_revenue_metrics_df.round(3)
```

## Evaluation 11 — Residual summary

```python
residual_summary = pd.DataFrame(
    {
        "Metric": [
            "RF Enrollment Mean Error",
            "RF Enrollment Mean Absolute Error",
            "GB Enrollment Mean Error",
            "GB Enrollment Mean Absolute Error",
            "RF Revenue Mean Error",
            "RF Revenue Mean Absolute Error",
            "GB Revenue Mean Error",
            "GB Revenue Mean Absolute Error"
        ],
        "Value": [
            prediction_evaluation["RF_Enrollment_Error"].mean(),
            prediction_evaluation["RF_Enrollment_Absolute_Error"].mean(),
            prediction_evaluation["GB_Enrollment_Error"].mean(),
            prediction_evaluation["GB_Enrollment_Absolute_Error"].mean(),
            prediction_evaluation["RF_Revenue_Error"].mean(),
            prediction_evaluation["RF_Revenue_Absolute_Error"].mean(),
            prediction_evaluation["GB_Revenue_Error"].mean(),
            prediction_evaluation["GB_Revenue_Absolute_Error"].mean()
        ]
    }
)

residual_summary.round(3)
```
