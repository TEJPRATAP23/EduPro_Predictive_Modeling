```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
day9_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data\EduPro_Day9_Train_Test_Splits.xlsx"
)

day10_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data\EduPro_Day10_Baseline_Models.xlsx"
)

print("Day 9 input file:")
print(day9_file_path)

print("\nDay 9 file exists:")
print(day9_file_path.exists())

print("\nDay 10 input file:")
print(day10_file_path)

print("\nDay 10 file exists:")
print(day10_file_path.exists())
```

## Load Day 9 datasets

```python
train_data = pd.read_excel(
    day9_file_path,
    sheet_name="Train_Data"
)

test_data = pd.read_excel(
    day9_file_path,
    sheet_name="Test_Data"
)

X_train = pd.read_excel(
    day9_file_path,
    sheet_name="X_Train"
)

X_test = pd.read_excel(
    day9_file_path,
    sheet_name="X_Test"
)

y_train = pd.read_excel(
    day9_file_path,
    sheet_name="Y_Train"
)

y_test = pd.read_excel(
    day9_file_path,
    sheet_name="Y_Test"
)

print("Day 9 datasets loaded successfully.")

print("\nTrain_Data:", train_data.shape)
print("Test_Data:", test_data.shape)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## Load Day 10 baseline results

```python
day10_metrics = pd.read_excel(
    day10_file_path,
    sheet_name="Baseline_Metrics"
)

day10_comparison = pd.read_excel(
    day10_file_path,
    sheet_name="Model_Comparison"
)

day10_predictions = pd.read_excel(
    day10_file_path,
    sheet_name="Predictions"
)

print("Day 10 results loaded successfully.")

print("\nDay 10 metrics:")
print(day10_metrics)

print("\nDay 10 model comparison:")
print(day10_comparison)
```

## Validate Day 11 inputs

```python
print("========== DAY 11 INPUT VALIDATION ==========")

print("\nTraining rows:")
print(len(X_train))

print("\nTesting rows:")
print(len(X_test))

print("\nTraining features:")
print(X_train.shape)

print("\nTesting features:")
print(X_test.shape)

print("\nTraining targets:")
print(y_train.shape)

print("\nTesting targets:")
print(y_test.shape)

print("\nTraining missing values:")
print(
    X_train.isnull().sum().sum()
    +
    y_train.isnull().sum().sum()
)

print("\nTesting missing values:")
print(
    X_test.isnull().sum().sum()
    +
    y_test.isnull().sum().sum()
)
```

## Define targets and features

```python
target_columns = [
    "EnrollmentCount",
    "CourseRevenue"
]

feature_columns = [
    "CourseCategory",
    "CourseType",
    "CourseLevel",
    "CoursePrice",
    "CourseDuration",
    "CourseRating",
    "TeacherRating",
    "YearsOfExperience",
    "Expertise"
]

print("Targets:")
print(target_columns)

print("\nFeatures:")
print(feature_columns)

print("\nNumber of features:")
print(len(feature_columns))
```

## Identify feature types

```python
categorical_features = [
    "CourseCategory",
    "CourseType",
    "CourseLevel",
    "Expertise"
]

numerical_features = [
    "CoursePrice",
    "CourseDuration",
    "CourseRating",
    "TeacherRating",
    "YearsOfExperience"
]

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

## Create preprocessing pipeline

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Advanced-model preprocessing pipeline created successfully.")
```

## Create Random Forest models

```python
random_forest_enrollment = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                max_depth=None,
                min_samples_split=2,
                min_samples_leaf=1
            )
        )
    ]
)

random_forest_revenue = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                max_depth=None,
                min_samples_split=2,
                min_samples_leaf=1
            )
        )
    ]
)

print("Random Forest models created successfully.")
```

## Train Random Forest for EnrollmentCount

```python
random_forest_enrollment.fit(
    X_train,
    y_train["EnrollmentCount"]
)

rf_enrollment_predictions = (
    random_forest_enrollment.predict(X_test)
)

print("Random Forest trained for EnrollmentCount.")

print("\nFirst 5 predictions:")
print(rf_enrollment_predictions[:5])
```

## Train Random Forest for CourseRevenue

```python
random_forest_revenue.fit(
    X_train,
    y_train["CourseRevenue"]
)

rf_revenue_predictions = (
    random_forest_revenue.predict(X_test)
)

print("Random Forest trained for CourseRevenue.")

print("\nFirst 5 predictions:")
print(rf_revenue_predictions[:5])
```

## Create Gradient Boosting models

```python
gradient_boosting_enrollment = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                learning_rate=0.05,
                max_depth=2,
                random_state=42
            )
        )
    ]
)

gradient_boosting_revenue = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                learning_rate=0.05,
                max_depth=2,
                random_state=42
            )
        )
    ]
)

print("Gradient Boosting models created successfully.")
```

## Train Gradient Boosting for EnrollmentCount

```python
gradient_boosting_enrollment.fit(
    X_train,
    y_train["EnrollmentCount"]
)

gb_enrollment_predictions = (
    gradient_boosting_enrollment.predict(X_test)
)

print("Gradient Boosting trained for EnrollmentCount.")

print("\nFirst 5 predictions:")
print(gb_enrollment_predictions[:5])
```

## Train Gradient Boosting for CourseRevenue

```python
gradient_boosting_revenue.fit(
    X_train,
    y_train["CourseRevenue"]
)

gb_revenue_predictions = (
    gradient_boosting_revenue.predict(X_test)
)

print("Gradient Boosting trained for CourseRevenue.")

print("\nFirst 5 predictions:")
print(gb_revenue_predictions[:5])
```

## Calculate advanced model metrics

```python
advanced_results = []

models_and_predictions = {
    "Random Forest": {
        "EnrollmentCount": rf_enrollment_predictions,
        "CourseRevenue": rf_revenue_predictions
    },
    "Gradient Boosting": {
        "EnrollmentCount": gb_enrollment_predictions,
        "CourseRevenue": gb_revenue_predictions
    }
}

for model_name, target_predictions in models_and_predictions.items():

    for target in target_columns:

        actual = y_test[target]

        predictions = target_predictions[target]

        mae = mean_absolute_error(
            actual,
            predictions
        )

        rmse = np.sqrt(
            mean_squared_error(
                actual,
                predictions
            )
        )

        r2 = r2_score(
            actual,
            predictions
        )

        advanced_results.append({
            "Target": target,
            "Model": model_name,
            "MAE": mae,
            "RMSE": rmse,
            "R2": r2
        })

advanced_results_df = pd.DataFrame(
    advanced_results
)

advanced_results_df
```

## Display advanced-model metrics

```python
formatted_advanced_results = (
    advanced_results_df.copy()
)

formatted_advanced_results["MAE"] = (
    formatted_advanced_results["MAE"].round(3)
)

formatted_advanced_results["RMSE"] = (
    formatted_advanced_results["RMSE"].round(3)
)

formatted_advanced_results["R2"] = (
    formatted_advanced_results["R2"].round(4)
)

formatted_advanced_results
```

## Combine Day 10 and Day 11 results

```python
all_model_results = pd.concat(
    [
        day10_metrics,
        advanced_results_df
    ],
    ignore_index=True
)

all_model_results
```

## Compare all models automatically

```python
print("========== ALL MODEL COMPARISON ==========")

for target in target_columns:

    target_results = (
        all_model_results[
            all_model_results["Target"] == target
        ]
        .sort_values(by="MAE")
        .reset_index(drop=True)
    )

    print(f"\nTarget: {target}")
    print(target_results)
```

## Identify the best model

```python
best_models = []

for target in target_columns:

    target_results = all_model_results[
        all_model_results["Target"] == target
    ].copy()

    best_row = target_results.loc[
        target_results["MAE"].idxmin()
    ]

    best_models.append({
        "Target": target,
        "Best_Model": best_row["Model"],
        "MAE": best_row["MAE"],
        "RMSE": best_row["RMSE"],
        "R2": best_row["R2"]
    })

best_models_df = pd.DataFrame(best_models)

best_models_df
```

## Create advanced prediction comparison

```python
advanced_prediction_comparison = test_data[
    ["CourseID", "CourseName"]
].copy()

advanced_prediction_comparison[
    "Actual_EnrollmentCount"
] = y_test["EnrollmentCount"].values

advanced_prediction_comparison[
    "Random_Forest_EnrollmentCount"
] = rf_enrollment_predictions

advanced_prediction_comparison[
    "Gradient_Boosting_EnrollmentCount"
] = gb_enrollment_predictions

advanced_prediction_comparison[
    "Actual_CourseRevenue"
] = y_test["CourseRevenue"].values

advanced_prediction_comparison[
    "Random_Forest_CourseRevenue"
] = rf_revenue_predictions

advanced_prediction_comparison[
    "Gradient_Boosting_CourseRevenue"
] = gb_revenue_predictions

advanced_prediction_comparison
```

## Round prediction results

```python
numeric_prediction_columns = [
    "Actual_EnrollmentCount",
    "Random_Forest_EnrollmentCount",
    "Gradient_Boosting_EnrollmentCount",
    "Actual_CourseRevenue",
    "Random_Forest_CourseRevenue",
    "Gradient_Boosting_CourseRevenue"
]

advanced_prediction_comparison = (
    advanced_prediction_comparison.round(
        {
            column: 2
            for column in numeric_prediction_columns
        }
    )
)

advanced_prediction_comparison
```
