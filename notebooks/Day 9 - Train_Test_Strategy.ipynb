```python
import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
day8_file_path = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\feature_data\EduPro_Day8_Prediction_Targets.xlsx"
)

print("Day 8 input file:")
print(day8_file_path)

print("File exists:", day8_file_path.exists())
```

```python
prediction_targets = pd.read_excel(
    day8_file_path,
    sheet_name="Prediction_Targets"
)

print("Day 8 dataset loaded successfully.")
print("Shape:", prediction_targets.shape)
```

## Inspect columns

```python
print("Available columns:")

for i, col in enumerate(prediction_targets.columns, start=1):
    print(f"{i}. {col}")
```

## Validating Day 8 input

```python
print("========== DAY 8 INPUT VALIDATION ==========")

print("\nDataset shape:")
print(prediction_targets.shape)

print("\nMissing values:")
print(prediction_targets.isnull().sum().sum())

print("\nDuplicate CourseIDs:")
print(prediction_targets["CourseID"].duplicated().sum())

print("\nUnique courses:")
print(prediction_targets["CourseID"].nunique())
```

## Define Prediction Targets

```python
target_columns = [
    "EnrollmentCount",
    "CourseRevenue"
]

print("Prediction targets:")
print(target_columns)
```

## Define columns to exclude from modeling

```python
identifier_columns = [
    "CourseID",
    "CourseName",
    "TeacherID",
    "TeacherName"
]

print("Identifier columns excluded from modeling:")
print(identifier_columns)
```

## Create the modeling features list

```python
feature_columns = [
    col
    for col in prediction_targets.columns
    if col not in identifier_columns + target_columns
]

print("Modeling features:")
print(feature_columns)

print("\nNumber of modeling features:")
print(len(feature_columns))
```

## Create X and y

```python
X = prediction_targets[feature_columns].copy()

y = prediction_targets[target_columns].copy()

print("X shape:")
print(X.shape)

print("\ny shape:")
print(y.shape)
```

```python
X.head()
```

```python
y.head()
```

## Check target distribution before splitting 

```python
print("========== TARGET DISTRIBUTION ==========")

print("\nEnrollmentCount:")
print(y["EnrollmentCount"].describe())

print("\nCourseRevenue:")
print(y["CourseRevenue"].describe())
```

## Important revenue check

```python
print("Zero EnrollmentCount:")
print((y["EnrollmentCount"] == 0).sum())

print("\nZero CourseRevenue:")
print((y["CourseRevenue"] == 0).sum())
```

## Create train/test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Train/Test split completed.")
```

## Check Split sizes

```python
print("========== TRAIN / TEST SHAPES ==========")

print("\nX_train:")
print(X_train.shape)

print("\nX_test:")
print(X_test.shape)

print("\ny_train:")
print(y_train.shape)

print("\ny_test:")
print(y_test.shape)
```

## Verify target distribution in train/test

```python
print("========== TRAIN TARGET DISTRIBUTION ==========")

print("\nEnrollmentCount:")
print(y_train["EnrollmentCount"].describe())

print("\nCourseRevenue:")
print(y_train["CourseRevenue"].describe())


print("\n========== TEST TARGET DISTRIBUTION ==========")

print("\nEnrollmentCount:")
print(y_test["EnrollmentCount"].describe())

print("\nCourseRevenue:")
print(y_test["CourseRevenue"].describe())
```

## Check zero-revenue representation

```python
print("Zero CourseRevenue in training set:")
print((y_train["CourseRevenue"] == 0).sum())

print("\nZero CourseRevenue in test set:")
print((y_test["CourseRevenue"] == 0).sum())
```

## Check for train/test overlap

```python
train_indices = X_train.index
test_indices = X_test.index

overlap = set(train_indices).intersection(set(test_indices))

print("Train/Test index overlap:")
print(len(overlap))
```

## Verify that all 60 observation are accounted for

```python
total_rows = len(X_train) + len(X_test)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Total rows:", total_rows)
print("Original rows:", len(X))
```

## Preserve CourseIDs for validation

```python
course_ids = prediction_targets["CourseID"].copy()

course_id_train = course_ids.loc[X_train.index].copy()
course_id_test = course_ids.loc[X_test.index].copy()

print("Training CourseIDs:", len(course_id_train))
print("Testing CourseIDs:", len(course_id_test))
```

## Create complete train/test datasets

```python
train_data = prediction_targets.loc[X_train.index].copy()
test_data = prediction_targets.loc[X_test.index].copy()

print("Training dataset:")
print(train_data.shape)

print("\nTesting dataset:")
print(test_data.shape)
```

## Output folder (Day 9)

```python
split_folder = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\model_data"
)

split_folder.mkdir(
    parents=True,
    exist_ok=True
)

print("Output folder:")
print(split_folder)

print("Folder exists:", split_folder.exists())
```

```python
day9_file_path = (
    split_folder
    / "EduPro_Day9_Train_Test_Splits.xlsx"
)

print("Day 9 output file:")
print(day9_file_path)
```

```python
with pd.ExcelWriter(
    day9_file_path,
    engine="openpyxl"
) as writer:

    train_data.to_excel(
        writer,
        sheet_name="Train_Data",
        index=False
    )

    test_data.to_excel(
        writer,
        sheet_name="Test_Data",
        index=False
    )

    X_train.to_excel(
        writer,
        sheet_name="X_Train",
        index=False
    )

    X_test.to_excel(
        writer,
        sheet_name="X_Test",
        index=False
    )

    y_train.to_excel(
        writer,
        sheet_name="Y_Train",
        index=False
    )

    y_test.to_excel(
        writer,
        sheet_name="Y_Test",
        index=False
    )

print("✅ Day 9 Train/Test datasets saved successfully:")
print(day9_file_path)
```

```python
check_train = pd.read_excel(
    day9_file_path,
    sheet_name="Train_Data"
)

check_test = pd.read_excel(
    day9_file_path,
    sheet_name="Test_Data"
)

print("Reloaded training dataset:")
print(check_train.shape)

print("\nReloaded testing dataset:")
print(check_test.shape)
```

## Day 9 validation

```python
print("========== DAY 9 TRAIN/TEST VALIDATION ==========")

print("\nOriginal dataset rows:")
print(len(prediction_targets))

print("\nTraining rows:")
print(len(train_data))

print("\nTesting rows:")
print(len(test_data))

print("\nTotal split rows:")
print(len(train_data) + len(test_data))

print("\nTraining shape:")
print(X_train.shape)

print("\nTesting shape:")
print(X_test.shape)

print("\nTraining target shape:")
print(y_train.shape)

print("\nTesting target shape:")
print(y_test.shape)

print("\nTrain/Test index overlap:")
print(len(overlap))

print("\nTraining missing values:")
print(X_train.isnull().sum().sum() + y_train.isnull().sum().sum())

print("\nTesting missing values:")
print(X_test.isnull().sum().sum() + y_test.isnull().sum().sum())

print("\nOutput file:")
print(day9_file_path)

print("\n==============================================")
print("DAY 9 COMPLETED")
print("==============================================")
```
