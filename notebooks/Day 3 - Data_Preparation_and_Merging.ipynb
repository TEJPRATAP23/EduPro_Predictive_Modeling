```python
import pandas as pd
import numpy as np
```

```python
cleaned_file_path = r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\cleaned_data\EduPro_Cleaned.xlsx"
```

```python
users = pd.read_excel(
    cleaned_file_path,
    sheet_name="Users"
)

teachers = pd.read_excel(
    cleaned_file_path,
    sheet_name="Teachers"
)

courses = pd.read_excel(
    cleaned_file_path,
    sheet_name="Courses"
)

transactions = pd.read_excel(
    cleaned_file_path,
    sheet_name="Transactions"
)
```

## Checking data before merging

```python
print("Users:", users.shape)
print("Teachers:", teachers.shape)
print("Courses:", courses.shape)
print("Transactions:", transactions.shape)
```

## Checking key columns

```python
print("Users columns:")
print(users.columns.tolist())

print("\nTeachers columns:")
print(teachers.columns.tolist())

print("\nCourses columns:")
print(courses.columns.tolist())

print("\nTransactions columns:")
print(transactions.columns.tolist())
```

```python
users.columns
```

## Merging Transaction with Users

```python
merged_data = transactions.merge(
    users,
    on="UserID",
    how="left"
)
```

```python
print(merged_data.shape)
```

## Checking first merge

```python
merged_data.head()
```

## Verify the first merge

```python
print("Rows after Users merge:", len(merged_data))
```

```python
print(merged_data["UserName"].isnull().sum())
```

## Merging with Courses

```python
merged_data = merged_data.merge(
    courses,
    on="CourseID",
    how="left"
)
```

```python
print(merged_data.shape)
```

```python
merged_data.head()
```

## Verify the Course merge

```python

print("Rows after Courses merge:", len(merged_data))
```

```python
print("Missing CourseName:",
      merged_data["CourseName"].isnull().sum())
```

## Merging with Teachers

```python
merged_data = merged_data.merge(
    teachers,
    on="TeacherID",
    how="left"
)
```

```python
print(merged_data.shape)
```

```python
merged_data.head()
```

## Verify the final merge

```python
print("Final rows:", len(merged_data))
print("Final columns:", len(merged_data.columns))
```

## Checking final columns

```python
print(merged_data.columns.tolist())
```

## Handling duplicates columns names

```python
merged_data = merged_data.rename(columns={
    "Age_x": "UserAge",
    "Gender_x": "UserGender",
    "Age_y": "TeacherAge",
    "Gender_y": "TeacherGender"
})
```

## Important predictive-modeling preparation

```python
merged_data["TransactionYear"] = merged_data["TransactionDate"].dt.year
merged_data["TransactionMonth"] = merged_data["TransactionDate"].dt.month
merged_data["TransactionDay"] = merged_data["TransactionDate"].dt.day
merged_data["TransactionDayOfWeek"] = merged_data["TransactionDate"].dt.dayofweek
```

```python
merged_data["TransactionQuarter"] = merged_data["TransactionDate"].dt.quarter
```

## Saving working

```python
from pathlib import Path

prepared_folder = Path(
    r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\prepared_data"
)

prepared_folder.mkdir(parents=True, exist_ok=True)

prepared_file_path = prepared_folder / "EduPro_Prepared.xlsx"
```
