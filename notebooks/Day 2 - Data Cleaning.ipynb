```python
users = pd.read_excel(file_path, sheet_name="Users")
teachers = pd.read_excel(file_path, sheet_name="Teachers")
courses = pd.read_excel(file_path, sheet_name="Courses")
transactions = pd.read_excel(file_path, sheet_name="Transactions")
```

## Making Copies

```python
users_clean = users.copy()
teachers_clean = teachers.copy()
courses_clean = courses.copy()
transactions_clean = transactions.copy()
```

```python
print("Original Users:", users.shape)
print("Clean Users:", users_clean.shape)

print("Original Teachers:", teachers.shape)
print("Clean Teachers:", teachers_clean.shape)

print("Original Courses:", courses.shape)
print("Clean Courses:", courses_clean.shape)

print("Original Transactions:", transactions.shape)
print("Clean Transactions:", transactions_clean.shape)
```

## Missing Value Check

```python
print("Users Missing Values")
print(users_clean.isnull().sum().sum())

print("\nTeachers Missing Values")
print(teachers_clean.isnull().sum().sum())

print("\nCourses Missing Values")
print(courses_clean.isnull().sum().sum())

print("\nTransactions Missing Values")
print(transactions_clean.isnull().sum().sum())
```

## Clean Text Fields

```python
text_columns_users = ["UserID", "UserName", "Gender", "Email"]

for col in text_columns_users:
    users_clean[col] = users_clean[col].str.strip()
```

```python
text_columns_teachers = [
    "TeacherID",
    "TeacherName",
    "Gender",
    "Expertise"
]

for col in text_columns_teachers:
    teachers_clean[col] = teachers_clean[col].str.strip()
```

```python
text_columns_courses = [
    "CourseID",
    "CourseName",
    "CourseCategory",
    "CourseType",
    "CourseLevel"
]

for col in text_columns_courses:
    courses_clean[col] = courses_clean[col].str.strip()
```

```python
text_columns_transactions = [
    "TransactionID",
    "UserID",
    "CourseID",
    "PaymentMethod",
    "TeacherID"
]

for col in text_columns_transactions:
    transactions_clean[col] = transactions_clean[col].str.strip()
```

## Saving the cleaned work

```python
from pathlib import Path
import pandas as pd

# Create the cleaned_data folder
cleaned_folder = Path(r"D:\Data Analytics Project\EduPro_Predictive_Modeling\data\cleaned_data")
cleaned_folder.mkdir(parents=True, exist_ok=True)

# File to save
cleaned_file_path = cleaned_folder / "EduPro_Cleaned.xlsx"

# Save all cleaned sheets
with pd.ExcelWriter(cleaned_file_path, engine="openpyxl") as writer:
    users_clean.to_excel(writer, sheet_name="Users", index=False)
    teachers_clean.to_excel(writer, sheet_name="Teachers", index=False)
    courses_clean.to_excel(writer, sheet_name="Courses", index=False)
    transactions_clean.to_excel(writer, sheet_name="Transactions", index=False)

print(f"Cleaned workbook saved at:\n{cleaned_file_path}")
```
